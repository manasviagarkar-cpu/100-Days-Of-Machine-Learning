```python
import matplotlib.pyplot as plt

# Function: f(x) = x^2 + 5x + 6
def f(x):
    return x**2 + 5*x + 6

# Derivative: f'(x) = 2x + 5
def derivative(x):
    return 2*x + 5

# Gradient descent function
def gradient_descent(start_x, learning_rate, steps):
    x = start_x
    history = []

    for step in range(steps):
        history.append(x)

        gradient = derivative(x)
        x = x - learning_rate * gradient

    return history

# Settings
start_x = 10
steps = 30
learning_rates = [0.01, 0.1, 0.5]

# Run gradient descent for each learning rate
results = {}

for lr in learning_rates:
    results[lr] = gradient_descent(start_x, lr, steps)

# Print tracked x values
for lr, values in results.items():
    print(f"\nLearning rate: {lr}")
    for step, x in enumerate(values):
        print(f"Step {step}: x = {x:.6f}, f(x) = {f(x):.6f}")

# Plot x values over time
plt.figure(figsize=(10, 6))

for lr, values in results.items():
    if lr == 0.01:
        label = "lr = 0.01 slow convergence"
    elif lr == 0.1:
        label = "lr = 0.1 good convergence"
    else:
        label = "lr = 0.5 fastest convergence"

    plt.plot(range(steps), values, marker='o', label=label)

# Show true minimum line
plt.axhline(y=-2.5, color='black', linestyle='--', label='Minimum x = -2.5')

plt.title("Gradient Descent on f(x) = x² + 5x + 6")
plt.xlabel("Step")
plt.ylabel("x value")
plt.legend()
plt.grid(True)
plt.show()
```
