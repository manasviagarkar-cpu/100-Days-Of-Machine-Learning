**MINI PROJECT**

```python
import time
import numpy as np
import matplotlib.pyplot as plt


# Matrix multiplication using nested loops
def multiply_with_loops(A, B):
    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])

    result = []

    for i in range(rows_A):
        row = []

        for j in range(cols_B):
            total = 0

            for k in range(cols_A):
                total += A[i][k] * B[k][j]

            row.append(total)

        result.append(row)

    return result


# Matrix multiplication using NumPy
def multiply_with_numpy(A, B):
    return np.dot(A, B)


# Matrix size
size = 100

# Create two random matrices
A_numpy = np.random.randint(1, 10, (size, size))
B_numpy = np.random.randint(1, 10, (size, size))

# Convert NumPy arrays into normal Python lists
A_list = A_numpy.tolist()
B_list = B_numpy.tolist()


# Time nested loop multiplication
start_time = time.time()
loop_result = multiply_with_loops(A_list, B_list)
loop_time = time.time() - start_time


# Time NumPy multiplication
start_time = time.time()
numpy_result = multiply_with_numpy(A_numpy, B_numpy)
numpy_time = time.time() - start_time


# Print the times
print("Nested loops time:", loop_time, "seconds")
print("NumPy time:", numpy_time, "seconds")

speed_difference = loop_time / numpy_time
print("NumPy is about", speed_difference, "times faster")


# Plot bar chart
methods = ["Nested Loops", "NumPy"]
times = [loop_time, numpy_time]

plt.bar(methods, times)
plt.ylabel("Time in seconds")
plt.title("Matrix Multiplication Speed Comparison")
plt.show()
```
