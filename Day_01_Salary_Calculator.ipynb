```python
name=input("Enter Employee Name:")
role = input("Enter Job Role: ")
hours_worked = float(input("Enter total hours worked: "))
hourly_rate = float(input("Enter hourly rate: "))
```

```python
gross_salary = hours_worked * hourly_rate
tax_amount = gross_salary * 0.10
net_salary = gross_salary - tax_amount
```

**Printing the Pay Slip**

```python
print(f"{'PAY SLIP':^30}\n")
print(f"Employee Name : {name}")
print(f"Job Role      : {role}\n")
print(f"Gross Salary  : ${gross_salary:,.2f}")
print(f"Tax (10%)     : -${tax_amount:,.2f}\n")
print(f"NET PAY       : ${net_salary:,.2f}")
```
