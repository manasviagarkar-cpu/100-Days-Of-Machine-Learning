Project 1: Dynamic Multiplication Table

```python
number = int(input("Enter a number for the table: ") or 5)

print(f"\nMultiplication Table for {number}:")
print("-" * 20)

# range(1, 11) starts at 1 and stops BEFORE 11
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i:2} = {result:3}")
```

Project 2: ML-FizzBuzz (Multiples of 3, 5, and 7)

```python
print("--- ML-FizzBuzz (1 to 50) ---")

for i in range(1, 51):
    output = ""
    if i % 3 == 0: output += "Fizz"
    if i % 5 == 0: output += "Buzz"
    if i % 7 == 0: output += "ML"

    print(output or i)
```

Project 3: The "One-Liner" (List Comprehension)

```python
# Logic: 'Fizz' if i%3==0 else '' ... combined into one list
ml_fizzbuzz = [
    ("Fizz"*(i%3==0) + "Buzz"*(i%5==0) + "ML"*(i%7==0)) or i
        for i in range(1, 51)
        ]

print(ml_fizzbuzz)
```
