```python
# Mini Project: Mean, Median, Mode, Variance, Standard Deviation
# Pure Python calculation + NumPy verification

import numpy as np

# Given list of 20 numbers
numbers = [12, 15, 18, 12, 20, 25, 30, 18, 12, 40,
           35, 22, 18, 50, 45, 30, 28, 12, 18, 20]

# ---------------- Pure Python Functions ----------------

def find_mean(data):
    total = sum(data)
    count = len(data)
    return total / count


def find_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    middle = n // 2

    # If number of values is even
    if n % 2 == 0:
        return (sorted_data[middle - 1] + sorted_data[middle]) / 2
    else:
        return sorted_data[middle]


def find_mode(data):
    frequency = {}

    # Count how many times each number appears
    for number in data:
        if number in frequency:
            frequency[number] += 1
        else:
            frequency[number] = 1

    highest_count = max(frequency.values())

    modes = []
    for number in frequency:
        if frequency[number] == highest_count:
            modes.append(number)

    return sorted(modes)


def find_variance(data):
    mean = find_mean(data)
    total = 0

    for number in data:
        total += (number - mean) ** 2

    return total / len(data)   # Population variance


def find_std_deviation(data):
    variance = find_variance(data)
    return variance ** 0.5


# ---------------- Pure Python Results ----------------

python_mean = find_mean(numbers)
python_median = find_median(numbers)
python_mode = find_mode(numbers)
python_variance = find_variance(numbers)
python_std = find_std_deviation(numbers)


# ---------------- NumPy Verification ----------------

numpy_array = np.array(numbers)

numpy_mean = np.mean(numpy_array)
numpy_median = np.median(numpy_array)

# NumPy mode calculation using unique values and counts
unique_values, counts = np.unique(numpy_array, return_counts=True)
highest_count = np.max(counts)
numpy_mode = sorted(unique_values[counts == highest_count].tolist())

numpy_variance = np.var(numpy_array)   # Population variance
numpy_std = np.std(numpy_array)


# ---------------- Side-by-Side Comparison ----------------

print("Given Numbers:")
print(numbers)

print("\nSide-by-Side Comparison")
print("-" * 60)
print(f"{'Measure':<20}{'Pure Python':<20}{'NumPy':<20}")
print("-" * 60)

print(f"{'Mean':<20}{python_mean:<20}{numpy_mean:<20}")
print(f"{'Median':<20}{python_median:<20}{numpy_median:<20}")
print(f"{'Mode':<20}{str(python_mode):<20}{str(numpy_mode):<20}")
print(f"{'Variance':<20}{python_variance:<20}{numpy_variance:<20}")
print(f"{'Std Deviation':<20}{python_std:<20}{numpy_std:<20}")

print("-" * 60)


# ---------------- Match Verification ----------------

print("\nVerification:")

print("Mean Match:", np.isclose(python_mean, numpy_mean))
print("Median Match:", np.isclose(python_median, numpy_median))
print("Mode Match:", python_mode == numpy_mode)
print("Variance Match:", np.isclose(python_variance, numpy_variance))
print("Std Deviation Match:", np.isclose(python_std, numpy_std))
```
