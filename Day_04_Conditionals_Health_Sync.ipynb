The "Health-Sync" Step **Analyzer**

```python
# Daily Step Data for one week
step_counts = [3200, 5100, 10200, 7800, 4500, 12000, 6000]

print("--- Health-Sync Daily Activity Report ---")
print(f"{'Day':<10} | {'Steps':<10} | {'Status'}")
print("-" * 35)

for i, steps in enumerate(step_counts, start=1):
      # Decision Logic
      if steps < 5000:
         status = "Sedentary 🛋️"
      elif steps <= 7499:
        status = "Lightly Active 🚶"
      elif steps <= 9999:
        status = "Moderately Active 🏃"
      else:
        status = "Very Active 🔥"

     # Ternary Operator Example: Quick check for a 'Goal Met' badge
      goal_badge = "⭐" if steps >= 10000 else ""
      print(f"Day {i:<6} | {steps:<10} | {status} {goal_badge}")
```
