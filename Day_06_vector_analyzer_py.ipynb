**Representing Vectors in Python**

```python
import numpy as np

x=np.array([1,2,3,4,5,6])

print(f"Vector v {x}")
print("vector index 0:",x[0])
print("vector index 3:",x[3])

# basci operation on vector
#add
x1=np.array([1,2,3])
x2=np.array([4,5,6])
print("x1",x1,"\nx2",x2)
xadd=x1+x2
print("vector add ",xadd)

#difference
xsub=x1-x2
print("vector difference ",xsub)

#multiplication
xmul=x1*2
print("vector multiplication",xmul)

# Calculating magnitude and direction
magnitude = np.linalg.norm(x1)
direction = np.arctan2(x1[0],x1[1])
print("Magnitude of x1:", magnitude)
print("Direction of x1:", direction, "radians")

# dot product
xdot=np.dot(x1,x2)
print("dot product",xdot)

# cross product
xcross=np.cross(x1,x2)
print("cross product", xcross)

# vector norm
L2_norm =np.linalg.norm(x1)
print("Vector norm", L2_norm)
```

**Vector Field Visualization**

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(6,-3,4)
y=np.linspace(5,-3,4)
X ,Y=np.meshgrid(x,y)
U = -Y
V = X
plt.quiver(X,Y,U,V)
plt.xlim(6,-3)
plt.ylim(5,-3)
plt.title("Circular Vector Field")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid()
plt.show()
```

**Mini project**

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.array([5,6])
y1 = np.array([9,2])

v_add = np.add(x1 , y1)
v_sub = np.subtract(x1,y1)
n = 5
v_scalar_mult = y1 * n

# Step 2: Dot product
dot_product = np.dot(x1 , y1)

# Visualization
plt.quiver(0, 0, x1[0], y1[1], angles='xy', scale_units='xy', scale=1, color='r')
plt.quiver(0, 0, x1[0], y1[1], angles='xy', scale_units='xy', scale=1, color='b')
plt.quiver(0, 0, v_add[0], v_add[1], angles='xy', scale_units='xy', scale=1, color='g')
plt.xlim(-2, 10)
plt.ylim(-2, 10)
plt.grid()
plt.title('Vector Operations: Addition (Green), V1 (Red), V2 (Blue)')
plt.show()

# Print results
print("Added Vectors:", v_add)
print("Subtracted Vectors:", v_sub)
print("Scalar Multiplication:", v_scalar_mult)
print("Dot Product:", dot_product)
```
