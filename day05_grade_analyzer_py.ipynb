The "Old School" Grade **Analyzer**

```python
import os

# 1. Create a dummy CSV file for testing
def create_test_csv(filename):
  data = "Name,Score\nAlice,85\nBob,72\nCharlie,98\nDavid,60\nEve,91"
  with open(filename, "w") as f:
    f.write(data)
```

```python
# 2. Function to read CSV manually
def read_csv(filename):
  students = []
  with open(filename, "r") as file:
    lines = file.readlines()
    # Skip the header (Name,Score) and loop through data
    for line in lines[1:]:
      name, score = line.strip().split(",")
      students.append({"name": name, "score": int(score)})
    return students
```

```python
# 3. Function to calculate statistics
def calculate_stats(students):
    scores = [s['score'] for s in students]
    avg = sum(scores) / len(scores)
    highest = max(students, key=lambda x: x['score'])
    lowest = min(students, key=lambda x: x['score'])
    return avg, highest, lowest
```

```python
# 4. Function to generate and save report
def print_report(avg, high, low, output_file):
    report = f"""--- GRADE SUMMARY REPORT ---
    Average Score: {avg:.2f}
    Top Performer: {high['name']} ({high['score']})
    Lowest Score:  {low['name']} ({low['score']})
    ----------------------------
    """
    print(report)
    with open(output_file, "w") as f:
      f.write(report)
```

```python
# --- EXECUTION ---
csv_name = "student_marks.csv"
output_name = "output.txt"

create_test_csv(csv_name) # Setup
data = read_csv(csv_name) # Step 1
average, top, bottom = calculate_stats(data) # Step 2
print_report(average, top, bottom, output_name) # Step 3
```
