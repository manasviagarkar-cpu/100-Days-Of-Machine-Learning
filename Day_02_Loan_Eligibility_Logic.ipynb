```python
print("--- SafePass AI Loan Eligibility Engine ---")
name = input("Customer Name: ")
age = int(input("Customer Age: ") or 0)  # Short-circuit fallback: if empty, use 0
credit_score = int(input("Credit Score (300-850): "))
monthly_income = float(input("Monthly Income ($): "))
has_bankruptcy = input("History of Bankruptcy? (True/False): ").lower() == 'true'
referred_by_manager = input("Do you have a Manager Referral? (True/False): ").lower() == 'true'
```

```python
# Rule 1: Age must be between 18 and 65 (Chained Comparison)
is_age_valid = 18 <= age <= 65

# Rule 2: High Credit Score (>= 750) OR (Moderate Score > 600 AND High Income > 5000)
is_financially_stable = (credit_score >= 750) or (credit_score > 600 and monthly_income > 5000)

# Rule 3: No Bankruptcy unless they have a Manager Referral (Logical Exception)
is_risk_acceptable = (not has_bankruptcy) or referred_by_manager

# Final Decision Gate
is_eligible = is_age_valid and is_financially_stable and is_risk_acceptable
```

```python
print(f"\nProcessing application for: {name}...\n")
print(f"Age Requirement Met:      {is_age_valid}")
print(f"Financial Stability Met: {is_financially_stable}\n")
print(f"Risk Assessment Passed:  {is_risk_acceptable}")

# Result Logic
if is_eligible:
    decision = "APPROVED ✅"
else:
      decision = "DENIED ❌"

      print(f"FINAL LOAN DECISION: {decision}")
```
